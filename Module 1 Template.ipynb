    * **Symptom medications:** Medications like donepezil, rivastigmine, galantamine, and memantine can help with memory and thinking symptoms, but they do not stop Alzheimer's from getting worse. ([Alzheimer's Association, 2026](https://doi.org/10.1002/alz.71345))

* **Disease progression & prognosis**
    * **Before symptoms:** Changes in the brain can start happening years before a person begins showing noticeable Alzheimer's symptoms. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00942-2))
    * **Getting worse over time:** Alzheimer's gets worse over time. Memory and thinking problems become more serious, and patients may eventually need help with everyday activities. ([Alzheimer's Association, 2026](https://doi.org/10.1002/alz.71345))
    * **Biomarkers:** Amyloid and tau biomarkers can help show what is happening in the brain and can give information about the stage of the disease. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00942-2))
    * **Different for each person:** Alzheimer's does not progress at exactly the same rate in everyone, so the amount of time it takes for symptoms to get worse can vary between patients. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00942-2))

* **Continuum of care providers**

## Data-Set: 
For this project, we are using data from the Seattle Alzheimer's Disease Brain Cell Atlas (SEA-AD). The dataset has information from 84 brain donors who had different levels of Alzheimer's disease. The donors were part of the Adult Changes in Thought (ACT) study and the University of Washington Alzheimer's Disease Research Center (UW ADRC).

The dataset includes basic patient information such as age at death, sex, APOE genotype, cognitive status, and age when cognitive symptoms started. It also includes cognitive test scores such as CASI, MMSE, and MOCA. There is also information about Alzheimer's disease in the brain, including Thal phase, Braak stage, and CERAD score.

After the donors passed away, their brain tissue was collected and studied. The researchers mainly looked at an area of the brain called the middle temporal gyrus (MTG). They used techniques such as immunohistochemistry to study Alzheimer's-related changes in the tissue. The larger SEA-AD study also used methods such as single-nucleus RNA sequencing (snRNA-seq), snATAC-seq, and MERFISH to study the brain tissue in more detail.

For our analysis, some of the main measurements we use are ABeta40, ABeta42, total tau (tTAU), and phosphorylated tau (pTAU). These protein measurements are given in pg/ug in our dataset. Age is measured in years, while CASI, MMSE, and MOCA are given as cognitive test scores.

#### Analysis 1: ABeta42 Levels in Female and Male Patients with Dementia

For my first analysis, I wanted to look at whether ABeta42 levels were different between female and male patients with dementia. I separated the patients into female and male groups and looked at the ABeta42 values for each group.

I found the average ABeta42 level for both groups and also calculated the standard deviation to see how spread out the values were. I then used a t-test to check if the difference between the female and male groups was significant. Finally, I made a bar graph to compare the average ABeta42 levels between the two groups and used error bars to show the standard deviation.





```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

from patient_Hira import Patient


# read the csv file
df = pd.read_csv("Metadata and Protein Data for Module 1.csv")


# make patient objects
patients = []

for index, row in df.iterrows():

    patient = Patient(
        row["Donor ID"],
        row["Age at Death"],
        row["Sex"],
        row["Cognitive Status"],
        row["APOE Genotype"],
        row["Thal"],
        row["ABeta40 pg/ug"],
        row["ABeta42 pg/ug"],
        row["tTAU pg/ug"],
        row["pTAU pg/ug"]
    )

    patients.append(patient)


print("Number of patients:", len(patients))


# get ABeta42 values for females and males with dementia
female_abeta = []
male_abeta = []

for patient in patients:

    if patient.sex == "Female" and patient.cognitive_status == "Dementia":
        if pd.notna(patient.abeta42):
            female_abeta.append(patient.abeta42)

    if patient.sex == "Male" and patient.cognitive_status == "Dementia":
        if pd.notna(patient.abeta42):
            male_abeta.append(patient.abeta42)


# find mean and standard deviation
female_mean = np.mean(female_abeta)
male_mean = np.mean(male_abeta)

female_std = np.std(female_abeta, ddof=1)
male_std = np.std(male_abeta, ddof=1)


# do t test
t_stat, p_val = stats.ttest_ind(
    female_abeta,
    male_abeta,
    equal_var=False
)

print("female mean =", female_mean)
print("male mean =", male_mean)
print("female standard deviation =", female_std)
print("male standard deviation =", male_std)
print("t-stat =", t_stat)
print("p-value =", p_val)


# make bar graph
groups = ["Female", "Male"]
means = [female_mean, male_mean]
stds = [female_std, male_std]

plt.bar(groups, means, yerr=stds, capsize=5)

plt.xlabel("Sex")
plt.ylabel("ABeta42 (pg/ug)")
plt.title("Mean ABeta42 in Patients with Dementia")

plt.text(
    0.5,
    0.95,
    f"t = {t_stat:.2f}\np = {p_val:.3e}",
    ha="center",
    va="top",
    transform=plt.gca().transAxes
)

plt.show()
```

#### Analysis 2: ABeta42 Levels and Age at Death

For my second analysis, I wanted to see if there was a relationship between a patient's age at death and their ABeta42 level. I used the age at death and ABeta42 values for each patient to make a scatter plot.

I also used linear regression to add a line of best fit and calculated the R-squared value to see how strongly age at death was related to ABeta42 levels.

```python
# get age and ABeta42 values
ages = []
abeta42 = []

for patient in patients:

    if pd.notna(patient.age_at_death) and pd.notna(patient.abeta42):
        ages.append(patient.age_at_death)
        abeta42.append(patient.abeta42)


# do linear regression
slope, intercept, r_value, p_value, std_err = stats.linregress(
    ages,
    abeta42
)

r_squared = r_value ** 2

print("slope =", slope)
print("R squared =", r_squared)
print("p-value =", p_value)


# make line of best fit
x_line = np.linspace(min(ages), max(ages), 100)
y_line = slope * x_line + intercept


# make scatter plot
plt.scatter(ages, abeta42)
plt.plot(x_line, y_line)

plt.xlabel("Age at Death")
plt.ylabel("ABeta42 (pg/ug)")
plt.title("ABeta42 vs Age at Death")

plt.text(
    0.05,
    0.95,
    f"R² = {r_squared:.3f}",
    transform=plt.gca().transAxes,
    va="top"
)

plt.show()
```
