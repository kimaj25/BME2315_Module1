## Disease Background: 

* **Prevalence & incidence**
   * **Prevalence:** Approximately 7.4 million Americans currently live with Alzheimer’s dementia ([Alzheimer's Association, 2026](https://doi.org/10.1002/alz.71345)).
   * **Incidence:** rate of new cases increases dramatically with age; in 2011, the average incidence in people ages 65-74 was 0.4% (1 in every 1,000 people), 3.2% for ages 75-84, and 7.6% for ages 85 and older. By 2050, the annual number of new cases is projected to double ([Alzheimer's Association, 2026](https://doi.org/10.1002/alz.71345)).

* **Economic burden**
   * According to UsAgainstAlzheimer’s, the total annual out-of-pocket payments in the U.S. for healthcare, long-term care, and hospice care for Alzheimer’s patients (and other dementias) are projected to reach $1.1 trillion by 2050, assuming no cure/treatment will be developed ([UsAgainstAlzheimer’s, 2026](https://www.usagainstalzheimers.org/alzheimers-disease-get-facts)).
   * Programs like Medicare and Medicaid only cover $175 billion (68%) of out-of-pocket healthcare costs. Medicaid payments are on average 23 times higher for those with Alzheimer’s compared to those without, and Medicare payments are three times greater on average. By 2050, Alzheimer’s is expected to increase Medicare and Medicaid costs by over 330%. More than $250 billion is spent annually in out-of-pocket healthcare for Alzheimer’s, which is more than 179 times the amount spent on finding a cure ([Alzheimer's Association, 2026](https://doi.org/10.1002/alz.71345)).

* **Risk factors (genetic, lifestyle)** 
   * **Non-Modifiable Risk Factors:**
     * **Age:** the greatest risk; majority of Alzheimer’s dementia patients are age 65 or older (late-onset).
     * **Genetics:** inheriting one or two copies of the e4 form of the apolipoprotein E (APOE) gene increases the risk of developing the disease.
     * Family History of Alzheimer’s dementia increases risk (revolving around genetics and shared non-genetic socioeconomic factors)
   * **Modifiable Risk Factors:**
     * Traumatic brain injury (TBI)
     * Smoking
     * Diet
     * Physical Activity
     * Education
     * Cognitive/social engagement
     * Hypertension
     * Cardiovascular Health
     * Sleep
     * Sensory loss
     * Diabetes
     * Air pollution
   * There is a consensus in Alzheimer’s research that suggests people who adopt healthy lifestyle habits can lower their risk/slow down the development of memory loss–the main habits being staying physically active and taking care/caution towards cardiovascular health ([Alzheimer's Association, 2026](https://doi.org/10.1002/alz.71345)).

* **Societal determinants**
   * Social determinants of health can influence both the risk of developing dementia and a person's ability to receive appropriate care. Factors such as income, education, housing, employment, access to health care, neighborhood safety, access to nutritious food, and environmental exposures can affect brain and cardiovascular health (e.g., limited access to healthcare can make it more difficult to identify and manage hypertension, diabetes, or other cardiovascular conditions associated with cognitive health). Socioeconomic conditions can also affect exposure to environmental risk factors such as air pollution and other potentially harmful substances. Furthermore, access to safe environments for physical activity, healthy food, social engagement, and educational opportunities can differ substantially between communities. Overall, research shows that social and environmental conditions can influence exposure to risk factors and access to protective resources ([Alzheimer's Association, 2026](https://doi.org/10.1002/alz.71345)).

* **Symptoms**
   * The key symptom of Alzheimer’s disease is memory loss. Early in the development of Alzheimer’s, people often have trouble remembering recent events or conversations. As the disease progresses, more symptoms appear. Since the brain changes from a physiological standpoint, factors of daily life including thinking and reasoning, judgement and decisions, planning and performing tasks, and personality and behavior can all be affected ([Mayo Clinic, 2024b](https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/symptoms-causes/syc-20350447)).

   * The following are common progressive symptoms of the disease: 
        * Difficulty remembering recent conversations or events
        * Difficulty completing familiar tasks
        * Problems with planning and problem-solving
        * Difficulty handling finances or paying bills
        * Poorer judgment
        * Difficulty finding or using words
        * Becoming disoriented to time or place
        * Misplacing objects and being unable to retrace one's steps
        * Changes in mood, personality, or behavior
        * Increasing difficulty performing activities of daily living

* **Diagnosis**
    * Diagnosing Alzheimer’s disease involves multiple types of assessments/criteria rather than one single test. Physicians typically begin by obtaining a medical history and asking about changes in memory, thinking, behavior, and the person's ability to perform everyday activities. Input from a family member or close friend can also be helpful because they may notice changes that the patient does not recognize ([National Institute on Aging, 2022](https://www.nia.nih.gov/health/alzheimers-symptoms-and-diagnosis/how-alzheimers-disease-diagnosed)).

    * The cognitive and neurological testing doctors may perform evaluate the following:
        * Memory
        * Attention
        * Language
        * Problem-solving
        * Counting and other thinking skills
        * Orientation
        * Ability to perform everyday activities
        * A physical and neurological examination may also evaluate reflexes, muscle strength, coordination, balance, vision, and hearing.

    * Brain imaging can be used to identify structural or functional changes and to rule out other causes of cognitive impairment.
        * **MRI:** Provides detailed images of brain structure and may show brain-region shrinkage or other abnormalities.
        * **CT:** Can help identify conditions such as tumors, strokes, bleeding, or head injuries.
        * **PET:** Different PET techniques can examine brain metabolism or detect Alzheimer’s-related biomarkers such as amyloid plaques or tau pathology ([Mayo Clinic, 2024a](https://www.nia.nih.gov/health/alzheimers-symptoms-and-diagnosis/how-alzheimers-disease-diagnosed)).

* **Standard of care treatments (& reimbursement)**
    * **Symptom medications:** Medications like donepezil, rivastigmine, galantamine, and memantine can help with memory and thinking symptoms, but they do not stop Alzheimer's from getting worse. ([Alzheimer's Association, 2026](https://doi.org/10.1002/alz.71345))
    * **Newer treatments:** Lecanemab and donanemab are newer treatments for people with early Alzheimer's. They target amyloid buildup in the brain and can help slow down the disease. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00942-2))
    * **Treatment monitoring:** Patients taking anti-amyloid treatments need to be monitored because these treatments are only used for certain patients and can have serious side effects. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00942-2))
    * **Medicare coverage:** Medicare Part B can cover FDA-approved anti-amyloid treatments for eligible patients. Patients usually pay 20% of the Medicare-approved amount after meeting their Part B deductible. ([Medicare, 2026](https://www.medicare.gov/coverage/monoclonal-antibodies-for-treating-early-alzheimers-disease))

* **Disease progression & prognosis**
    * **Before symptoms:** Changes in the brain can start happening years before a person begins showing noticeable Alzheimer's symptoms. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00942-2))
    * **Getting worse over time:** Alzheimer's gets worse over time. Memory and thinking problems become more serious, and patients may eventually need help with everyday activities. ([Alzheimer's Association, 2026](https://doi.org/10.1002/alz.71345))
    * **Biomarkers:** Amyloid and tau biomarkers can help show what is happening in the brain and can give information about the stage of the disease. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00942-2))
    * **Different for each person:** Alzheimer's does not progress at exactly the same rate in everyone, so the amount of time it takes for symptoms to get worse can vary between patients. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00942-2))

* **Continuum of care providers**
    * **Primary care:** Primary-care doctors can notice signs of memory problems, do early assessments, and refer patients to specialists when needed. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00989-1))
    * **Specialists:** Neurologists, geriatricians, and memory specialists can help diagnose Alzheimer's, choose treatments, and follow the patient as the disease progresses.
    * **Supportive care:** Nurses, physical and occupational therapists, social workers, and home-health workers can help patients with safety and everyday activities.
    * **Long-term care:** As Alzheimer's becomes more severe, some patients may need more help from family caregivers, home-care services, assisted living, or long-term care facilities. ([Alzheimer's Association, 2026](https://doi.org/10.1002/alz.71345))

* **Biological mechanisms**
    * **Amyloid:** Beta-amyloid (Aβ) can build up between brain cells and form plaques. This is one of the main changes seen in Alzheimer's disease. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00942-2))
    * **Tau:** Tau is normally found inside neurons, but in Alzheimer's it can become abnormal and build up inside the cells. This can affect how the neurons work. ([Nature Reviews Molecular Cell Biology, 2024](https://www.nature.com/articles/s41580-024-00753-9))
    * **Inflammation:** Alzheimer's can also cause changes in the brain's immune response. Long-term inflammation can contribute to damage in the brain. ([Nature Reviews Immunology, 2025](https://www.nature.com/articles/s41577-024-01104-7))
    * **Biomarkers:** Aβ42 and phosphorylated tau can be measured as biomarkers to help identify biological changes related to Alzheimer's disease. ([Nature Reviews Neurology, 2024](https://www.nature.com/articles/s41582-024-00989-1))

* **Clinical trials / next-generation therapies**
    * **New drugs:** There are many new Alzheimer's treatments being studied. In 2025, there were 138 drugs being tested across 182 clinical trials. ([Alzheimer's & Dementia, 2025](https://doi.org/10.1002/trc2.70098))
    * **Different targets:** Researchers are not only studying amyloid. New treatments are also being developed for other processes involved in Alzheimer's disease. ([Alzheimer's & Dementia, 2025](https://doi.org/10.1002/trc2.70098))
    * **Tau treatments:** Researchers are also studying treatments that target abnormal tau because tau buildup is an important part of Alzheimer's disease. ([Nature Reviews Molecular Cell Biology, 2024](https://www.nature.com/articles/s41580-024-00753-9))
    * **Using biomarkers:** Biomarkers are being used in clinical trials to help choose patients for studies and see whether a new treatment is causing biological changes. ([Alzheimer's & Dementia, 2025](https://doi.org/10.1002/trc2.70098))

## Data Analysis: 

#### Analysis 1: ABeta42 Levels in Female and Male Patients with Dementia

For our first analysis, we wanted to look at whether Aβ42 levels were different between female and male patients with dementia. We separated the patients into female and male groups and looked at the Aβ42 values for each group.

We found the average Aβ42 level for both groups and also calculated the standard deviation to see how spread out the values were. We then used a t-test to check if the difference between the female and male groups was significant. Finally, we made a bar graph to compare the average ABeta42 levels between the two groups and used error bars to show the standard deviation.





```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

from patient_Hira import Patient


# read the csv file
df = pd.read_csv("Metadata and Protein Data for Module 1.csv")


# make patient objects
patients = []

for index, row in df.iterrows():

    patient = Patient(
        row["Donor ID"],
        row["Age at Death"],
        row["Sex"],
        row["Cognitive Status"],
        row["APOE Genotype"],
        row["Thal"],
        row["ABeta40 pg/ug"],
        row["ABeta42 pg/ug"],
        row["tTAU pg/ug"],
        row["pTAU pg/ug"]
    )

    patients.append(patient)


print("Number of patients:", len(patients))


# get ABeta42 values for females and males with dementia
female_abeta = []
male_abeta = []

for patient in patients:

    if patient.sex == "Female" and patient.cognitive_status == "Dementia":
        if pd.notna(patient.abeta42):
            female_abeta.append(patient.abeta42)

    if patient.sex == "Male" and patient.cognitive_status == "Dementia":
        if pd.notna(patient.abeta42):
            male_abeta.append(patient.abeta42)


# find mean and standard deviation
female_mean = np.mean(female_abeta)
male_mean = np.mean(male_abeta)

female_std = np.std(female_abeta, ddof=1)
male_std = np.std(male_abeta, ddof=1)


# do t test
t_stat, p_val = stats.ttest_ind(
    female_abeta,
    male_abeta,
    equal_var=False
)

print("female mean =", female_mean)
print("male mean =", male_mean)
print("female standard deviation =", female_std)
print("male standard deviation =", male_std)
print("t-stat =", t_stat)
print("p-value =", p_val)


# make bar graph
groups = ["Female", "Male"]
means = [female_mean, male_mean]
stds = [female_std, male_std]

plt.bar(groups, means, yerr=stds, capsize=5)

plt.xlabel("Sex")
plt.ylabel("ABeta42 (pg/ug)")
plt.title("Mean ABeta42 in Patients with Dementia")

plt.text(
    0.5,
    0.95,
    f"t = {t_stat:.2f}\np = {p_val:.3e}",
    ha="center",
    va="top",
    transform=plt.gca().transAxes
)

plt.show()
```

#### Analysis 2: ABeta42 Levels and Age at Death

For our second analysis, we wanted to see if there was a relationship between a patient's age at death and their Aβ42 level. We used the age at death and Aβ42 values for each patient to make a scatter plot.

We also used linear regression to add a line of best fit and calculated the R-squared value to see how strongly age at death was related to Aβ42 levels.

```python
# get age and ABeta42 values
ages = []
abeta42 = []

for patient in patients:

    if pd.notna(patient.age_at_death) and pd.notna(patient.abeta42):
        ages.append(patient.age_at_death)
        abeta42.append(patient.abeta42)


# do linear regression
slope, intercept, r_value, p_value, std_err = stats.linregress(
    ages,
    abeta42
)

r_squared = r_value ** 2

print("slope =", slope)
print("R squared =", r_squared)
print("p-value =", p_value)


# make line of best fit
x_line = np.linspace(min(ages), max(ages), 100)
y_line = slope * x_line + intercept


# make scatter plot
plt.scatter(ages, abeta42)
plt.plot(x_line, y_line)

plt.xlabel("Age at Death")
plt.ylabel("ABeta42 (pg/ug)")
plt.title("ABeta42 vs Age at Death")

plt.text(
    0.05,
    0.95,
    f"R² = {r_squared:.3f}",
    transform=plt.gca().transAxes,
    va="top"
)

plt.show()
```

#### Analysis 3: Age of Onset of Cognitive Symptoms in Female and Male Patients with Dementia

For our third analysis, we wanted to compare the mean age of onset of cognitive symptoms, as outlined in the background, in dementia patients by their sex to determine if gender is a factor in the incidence/prevelance of the disease. 

First, patients with dementia were separated into female and male groups. Age of onset of cognitive symptoms was collected for each group. The mean and standard deviation were calculated for each sex, and the means were compared using an independent two-sample Student's t-test. The results were visualized using a bar graph with standard-deviation error bars. The resulting means were 82.2 years for females and 78.5 years for males. The Student's t-test produced a p-value of about 0.4335. With a p-value greater than 0.05, the difference in the age of onset of cognitive symptoms between female and males is not statisitcally significant. Thus, we cannot conclude that sex is a determinant of the age of onset of cognitive symptoms among dementia patients.

#### Analysis 4-6: ABeta42 Levels and Different Cognitive Tests

For our last three analysis', we wanted to find if high Aβ42 levels corresponded with low performance on cognitive exam scores, and if different cognitive exams exhibited similar trends. A linear regression was performed separately for Last CASI score, Last MMSE score, and Last MoCA score against Aβ42 concentration. For each regression, the slope and coefficient of determination (R²) were calculated. The slope describes the direction and magnitude of the relationship between the cognitive score and Aβ42, while R² represents the proportion of variation in Aβ42 levels explained by the cognitive score.

For Last CASI score versus Aβ42, the regression equation was y = -5.228x + 518.828, with an R² value of 0.076. The negative slope indicates that higher CASI scores were associated with lower Aβ42 levels in this dataset. However, the relatively small R² indicates that CASI score explained only about 7.6% of the variation in Aβ42 levels.

As mentioned, the same process was repeated for the Last MMSE score versus Aβ42; the regression equation was y = -4.860x + 185.592, with an R² value of 0.018. This was also a negative relationship, but the R² indicates that MMSE score explained only about 1.8% of the variation in Aβ42 levels. Thus, there was very little linear association between MMSE score and Aβ42 in this sample.

Finally, for Last MOCA score versus Aβ42, the regression equation was y = -3.517x + 103.560, with an R² value of 0.278. This regression had the largest R² of the three cognitive measures, meaning that MOCA score explained approximately 27.8% of the variation in Aβ42 levels. Although this represents a stronger linear association than the CASI and MMSE analyses, most of the variation in Aβ42 levels remained unexplained by MOCA score alone.

Overall, all three regression slopes were negative, indicating higher cognitive exam scores correlate twith lower Aβ42 levels, but the strength of the relationships varied. MOCA had the highest R², followed by CASI and then MMSE. These results indicate that cognitive test scores alone did not account for most of the variation in Aβ42 concentrations within this dataset.
